# Random Survival Forest

### Wstęp
&emsp;Klasyczny model proporcjnoalnego hazardu Coxa okazał się nieskuteczny. Wynika to ze złamania podstawowego załozenia modelu o stałym wpływie cech w czasie.
Żeby poradzić sobie z tymi ograniczeniami wykorzystamy bardziej zaawansowany model nieliniowy: `Random Survival Forest`

### Biblioteki i przygotowanie danych
&emsp;Dane przygotowujemy tak, zeby odpowiadaly tym wykorzystanym przy klasycznym modelu Coxa.

```python
import pandas as pd
import numpy as np
from sksurv.ensemble import RandomSurvivalForest

df = pd.read_csv('../data/dataset_survclean.csv')
df['document_create_date_dt'] = pd.to_datetime(df['document_create_date_dt'])
df = df.sort_values('document_create_date_dt')

features = [
    'total_open_amount',
    'days_to_due',
    'invoice_age',
    'avg_delay_customer',
    'cust_payment_terms',
    'invoice_currency',
    'segment',
    'time_days', 
    'event'
]

rsf_df = df[features].copy()

rsf_df = pd.get_dummies(rsf_df, columns=['cust_payment_terms', 'invoice_currency', 'segment'], drop_first=True)
rsf_df = rsf_df.dropna()
rsf_df = rsf_df.astype(float)

split_id = int(len(rsf_df) * 0.8)
train_df = rsf_df.iloc[:split_id]
test_df = rsf_df[split_id:]

print(f"Zbiór treningowy: {len(train_df)} faktur\nZbiór testowy: {len(test_df)} faktur")

display(train_df.head())
```
